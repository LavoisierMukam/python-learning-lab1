## Question 4

I count the number of occurrences of each character in the string `"HelLo WorLd!!"`, while treating uppercase and lowercase letters as identical.

```python
s = "HelLo WorLd!!".lower()
counts = {}

for char in s:
    if char in counts:
        counts[char] += 1
    else:
        counts[char] = 1

print(counts)
```

# Part 2 — Fast computations with NumPy

## Question 1 — Comparing floating-point numbers

I compute `0.1 + 100 - 100`, then compare the result to `0.1` using both `==` and `np.allclose`.

```python
x = 0.1 + 100 - 100

print(x)
print(x == 0.1)
print(np.allclose(x, 0.1))
```

Direct comparison with `==` may fail for floating-point numbers because of numerical approximation.  
`np.allclose` is more appropriate when comparing floating-point values.

## Question 2 — For-loop versus NumPy

### 2.1 Using a for-loop

```python
squares_list = []

for i in range(1, 12):
    squares_list.append(i**2)

print(squares_list)
```

### 2.2 Using NumPy

```python
squares_array = np.arange(1, 12) ** 2
print(squares_array)
```

### 2.3 Comment

NumPy arrays should generally be preferred over Python loops for numerical computations because vectorized operations are faster, more concise, and easier to read.

## Question 4 — Vectorizing a product

I approximate the infinite product

\[
2\prod_{k=1}^{\infty}\frac{4k^2}{4k^2 - 1}
\]

by replacing $\infty$ with a large integer $n$.

### 4.1 Version using a for-loop

```python
n = 10000
prod_value = 1.0

for k in range(1, n + 1):
    prod_value *= (4 * k**2) / (4 * k**2 - 1)

result_loop = 2 * prod_value
print(result_loop)
```

### 4.2 Vectorized NumPy version

```python
k = np.arange(1, n + 1)
result_numpy = 2 * np.prod((4 * k**2) / (4 * k**2 - 1))

print(result_numpy)
```

### 4.3 Timing both versions

```python
start = time.time()

prod_value = 1.0
for k in range(1, n + 1):
    prod_value *= (4 * k**2) / (4 * k**2 - 1)
result_loop = 2 * prod_value

loop_time = time.time() - start

start = time.time()

k = np.arange(1, n + 1)
result_numpy = 2 * np.prod((4 * k**2) / (4 * k**2 - 1))

numpy_time = time.time() - start

print(f"Loop result  : {result_loop:.10f}")
print(f"NumPy result : {result_numpy:.10f}")
print(f"Loop time    : {loop_time:.6f} s")
print(f"NumPy time   : {numpy_time:.6f} s")
print(np.allclose(result_loop, result_numpy))
```

### 4.4 `%timeit`

In a Jupyter notebook, `%timeit` is more reliable than `time.time()` because it repeats the computation several times and gives a more stable estimate.

```python
%timeit 2 * np.prod((4 * np.arange(1, n + 1)**2) / (4 * np.arange(1, n + 1)**2 - 1))
```

## Question 9 — Linear algebra

### 9.1 Random matrix

```python
M = rng.uniform(-1, 1, size=(5, 10))
```

### 9.2 Define $G = M^T M$ and study its properties

```python
G = M.T @ M
print(G.shape)
print(np.allclose(G, G.T))
```

`G` is symmetric because it is of the form $M^T M$.

### Eigenvalues

```python
eigvals = np.linalg.eigvalsh(G)
print(eigvals)
```

### Positive semi-definiteness

```python
print(np.all(eigvals >= -1e-12))
```

This confirms numerically that $G$ is positive semi-definite.

### Strict positivity of eigenvalues

```python
print(np.all(eigvals > 1e-12))
```

This is generally `False` here, because `M` has shape `(5, 10)`, so the rank of $G$ is at most 5. Therefore, $G$ cannot have 10 strictly positive eigenvalues.

### 9.3 Rank

```python
rank_G = np.linalg.matrix_rank(G)
print(rank_G)
```

### 9.4 Euclidean norm

```python
fro_norm = np.linalg.norm(G)
print(fro_norm)
```

### 9.5 Operator norm

```python
op_norm = np.linalg.norm(G, 2)
print(op_norm)
```

### 9.6 Standard deviation of each column

```python
std_cols = G.std(axis=0)
print(std_cols)
```
